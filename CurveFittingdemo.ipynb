```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import scipy.optimize as fit
```

```python
df = pd.read_csv('S25_cf_demo_data.csv')
df.head()
```

```python
plt.figure()
plt.scatter(df['x'], df['y'])
plt.show()
```

```python
def f(x, m, b): #y = mx+b for straight line
    return m * x + b

p0 = [1,1] #giving m, b

params, cov = fit.curve_fit(f, df['x'], df['y'], p0)
```

```python
params #gives m,b
```

```python
cov 
```

```python
m = params[0]
b = params[1]

y_estimate = f(df['x'], m, b)
```

```python
plt.figure()
plt.plot(df['x'], y_estimate)
plt.scatter(df['x'], df['y'])
plt.show()
```

```python
#RESIDUALS
#difference between predicted and actual
residuals = y_estimate - df['y']

plt.figure()
plt.scatter(y_estimate, residuals)
plt.show #since it is not a cloud we know a straight line was wrong
```
