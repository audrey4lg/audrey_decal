```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.arange(0,100, 0.1) #default step is 1 so specify to make more specific
y = np.sin(x)
```

```python
plt.figure()
plt.plot(x, y, label='1')
plt.plot(x, y*0.5, label='0.5')
plt.xlabel('x')
plt.ylabel('y')
plt.title('sine')
plt.legend()
plt.show()
```

```python
#CUSTOMIZATION
# scatter plot: plt.scatter(x, y)
# bar plot: plt.bar(x,y)
# change color: plt.plot(x, y, color='red')
#change size: plt.plot (x, y, linewidth = '20.5')
#marker: plt.plot(x,y,marker='*')
#opacity: plt.plot(x,y,alpha='0.5')
#size of whole plot: plt.figure(figsize=(12,8))
```

```python
x = np.linspace(1,10,10)
y = np.log(x)

plt.figure(figsize=(6, 4))
plt.scatter(x, y, color='olivedrab', alpha=.75, marker='*', s=100)
plt.plot(x, y, color='cornflowerblue', linewidth=3, linestyle='dashdot')
plt.title('shooting star')
plt.xlabel('ground')
plt.ylabel('sky')
plt.show
```

```python
fig, axs = plt.subplots(2,2) #() gives rows and colums

axs[0,0].plot(x, y, color='red')
axs[0,1].plot(x, y/2, color='orange')
axs[1,0].plot(x, y/2)
axs[1,1].plot(x, y/2, color='purple')

plt.show()
```
